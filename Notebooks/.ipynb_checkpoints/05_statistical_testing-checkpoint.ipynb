```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from scipy import stats
listings = pd.read_csv('../data/listings.csv.gz', compression='gzip', low_memory=False)
calendar = pd.read_csv('../data/calendar.csv.gz', compression='gzip')
reviews = pd.read_csv('../data/reviews.csv')
listings_clean = pd.read_csv('../data/listings_clean.csv')
```

```python
# 1. Clean the price column (removes '$' or commas if they exist, and turns it into numeric values)
listings_clean['price'] = pd.to_numeric(
    listings_clean['price'].astype(str).str.replace(r'[$,]', '', regex=True), 
    errors='coerce'
)

superhost = listings_clean[listings_clean['host_is_superhost'] == 't']['price']
regular = listings_clean[listings_clean['host_is_superhost'] == 'f']['price']

# 3. Run the t-test while explicitly dropping missing (NaN) values
t_stat, p_value = stats.ttest_ind(
    superhost.dropna(), 
    regular.dropna(), 
    equal_var=False)

print(f't = {t_stat:.2f}, p = {p_value:.4f}')
```
